# 🔢 PARTE 1: NumPy — Arreglos y Métodos

NumPy (Numerical Python) es la base del ecosistema científico de Python. Permite trabajar con arreglos multidimensionales de manera eficiente.

### 2.1 Creación de Arreglos

```python
# np.array() — Crear arreglo desde una lista
# Ejemplo: temperaturas diarias en °C de una semana en Bucaramanga
temperaturas = np.array([28.5, 30.1, 27.3, 29.8, 31.0, 26.5, 28.9])
print('Temperaturas semanales (°C):', temperaturas)
print('Tipo de dato:', temperaturas.dtype)
print('Dimensiones:', temperaturas.ndim)
print('Forma (shape):', temperaturas.shape)
```

```python
# np.zeros() — Arreglo de ceros
# Útil para inicializar matrices de pesos en ML
pesos_iniciales = np.zeros((3, 4))
print('Matriz de pesos inicializada en cero:')
print(pesos_iniciales)

# np.ones() — Arreglo de unos
# Útil para crear máscaras binarias
mascara = np.ones((2, 5), dtype=int)
print('\nMáscara binaria (unos):')
print(mascara)
```

```python
# np.arange() — Secuencia con paso definido
# Ejemplo: índices de tiempo en milisegundos
tiempo_ms = np.arange(0, 100, 10)
print('Instantes de tiempo (ms):', tiempo_ms)

# np.linspace() — N puntos equiespaciados en un rango
# Ejemplo: 8 puntos de muestreo entre 0 y 1
muestras = np.linspace(0, 1, 8)
print('Puntos de muestreo:', np.round(muestras, 3))
```

```python
# np.random.seed() + np.random.randn() — Datos aleatorios reproducibles
# Ejemplo: simular calificaciones de 20 estudiantes
np.random.seed(42)
calificaciones = np.random.randint(30, 100, size=20)
print('Calificaciones simuladas (20 estudiantes):')
print(calificaciones)

# np.random.normal() — Distribución normal
# Ejemplo: errores de medición con media 0 y std 0.5
errores = np.random.normal(loc=0, scale=0.5, size=10)
print('\nErrores de medición (distribución normal):')
print(np.round(errores, 4))
```

```python
# np.eye() — Matriz identidad
identidad = np.eye(4)
print('Matriz identidad 4x4:')
print(identidad)

# np.full() — Arreglo con valor constante
# Ejemplo: inicializar tabla de precios base
precios_base = np.full((3, 3), 15000)
print('\nTabla de precios base (COP):')
print(precios_base)
```

### 2.2 Manipulación de Arreglos

```python
# np.reshape() — Cambiar forma sin alterar datos
# Ejemplo: reorganizar 12 mediciones en una matriz 3x4 (3 sensores, 4 lecturas)
mediciones = np.arange(1, 13)
matriz_sensores = mediciones.reshape(3, 4)
print('Mediciones planas:', mediciones)
print('\nReorganizado como 3 sensores x 4 lecturas:')
print(matriz_sensores)
```

```python
# .flatten() — Aplanar a 1D (crea copia)
# .ravel() — Aplanar a 1D (vista si es posible, más eficiente)
aplanado = matriz_sensores.flatten()
print('flatten():', aplanado)

raveled = matriz_sensores.ravel()
print('ravel(): ', raveled)
```

```python
# .T / np.transpose() — Transponer matriz
# Muy usado en álgebra lineal para ML
print('Matriz original (3x4):')
print(matriz_sensores)
print('\nTranspuesta (4x3):')
print(matriz_sensores.T)
```

```python
# np.concatenate() — Unir arreglos a lo largo de un eje
lote1 = np.array([10, 20, 30])
lote2 = np.array([40, 50, 60])
total = np.concatenate([lote1, lote2])
print('Concatenación 1D:', total)

# np.vstack() — Apilar verticalmente (filas)
# np.hstack() — Apilar horizontalmente (columnas)
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print('\nvstack (apilado vertical):')
print(np.vstack([A, B]))
print('\nhstack (apilado horizontal):')
print(np.hstack([A, B]))
```

```python
# np.split() — Dividir arreglo en partes iguales
# Ejemplo: dividir dataset en train/val/test
datos_completos = np.arange(30)
train, val, test = np.split(datos_completos, [18, 24])  # 60% / 20% / 20%
print(f'Train ({len(train)} muestras):', train)
print(f'Val   ({len(val)} muestras):', val)
print(f'Test  ({len(test)} muestras):', test)
```

### 2.3 Métodos Estadísticos y Matemáticos

```python
# Estadísticas descriptivas con NumPy
np.random.seed(7)
ventas_diarias = np.random.randint(50, 500, size=30)  # 30 días de ventas

print('=== ESTADÍSTICAS DE VENTAS DIARIAS ===')
print(f'np.mean()   — Promedio   : ${np.mean(ventas_diarias):.1f}')
print(f'np.median() — Mediana    : ${np.median(ventas_diarias):.1f}')
print(f'np.std()    — Desv. Est. : ${np.std(ventas_diarias):.1f}')
print(f'np.var()    — Varianza   : ${np.var(ventas_diarias):.1f}')
print(f'np.min()    — Mínimo     : ${np.min(ventas_diarias)}')
print(f'np.max()    — Máximo     : ${np.max(ventas_diarias)}')
print(f'np.sum()    — Total      : ${np.sum(ventas_diarias)}')
print(f'np.ptp()    — Rango      : ${np.ptp(ventas_diarias)}')
```

```python
# np.percentile() / np.quantile() — Percentiles y cuantiles
# Muy útiles para detección de outliers (método IQR)
Q1 = np.percentile(ventas_diarias, 25)
Q3 = np.percentile(ventas_diarias, 75)
IQR = Q3 - Q1
limite_inf = Q1 - 1.5 * IQR
limite_sup = Q3 + 1.5 * IQR

print('=== DETECCIÓN DE OUTLIERS (Método IQR) ===')
print(f'Q1 (percentil 25): {Q1}')
print(f'Q3 (percentil 75): {Q3}')
print(f'IQR: {IQR}')
print(f'Límite inferior: {limite_inf:.1f}')
print(f'Límite superior: {limite_sup:.1f}')

outliers = ventas_diarias[(ventas_diarias < limite_inf) | (ventas_diarias > limite_sup)]
print(f'Valores atípicos encontrados: {outliers}')
```

```python
# np.argmin() / np.argmax() — Índice del mínimo/máximo
# Útil para encontrar qué día fue el mejor/peor en ventas
dias = [f'Día {i+1}' for i in range(30)]

idx_min = np.argmin(ventas_diarias)
idx_max = np.argmax(ventas_diarias)

print(f'Menor venta: ${ventas_diarias[idx_min]} → {dias[idx_min]}')
print(f'Mayor venta: ${ventas_diarias[idx_max]} → {dias[idx_max]}')
```

```python
# np.sort() — Ordenar arreglo
# np.argsort() — Índices que ordenarían el arreglo
puntajes = np.array([85, 92, 71, 88, 95, 60, 77])
estudiantes = ['Ana', 'Luis', 'María', 'Carlos', 'Sofia', 'Pedro', 'Juliana']

orden = np.argsort(puntajes)[::-1]  # Descendente
print('=== RANKING DE ESTUDIANTES ===')
for i, idx in enumerate(orden):
    print(f'{i+1}. {estudiantes[idx]}: {puntajes[idx]} puntos')
```

```python
# np.unique() — Valores únicos y sus conteos
# Ejemplo: categorías de productos vendidos
categorias_ventas = np.array(['Electrónica', 'Ropa', 'Electrónica', 'Alimentos',
                               'Ropa', 'Electrónica', 'Alimentos', 'Alimentos', 'Ropa'])

valores, conteos = np.unique(categorias_ventas, return_counts=True)
print('=== FRECUENCIA POR CATEGORÍA ===')
for cat, cnt in zip(valores, conteos):
    print(f'{cat}: {cnt} ventas')
```

```python
# np.where() — Condicional vectorizado (equivale a if-else por elemento)
# Ejemplo: etiquetar ventas como 'Alta' o 'Baja' según umbral
umbral = 250
etiquetas = np.where(ventas_diarias >= umbral, 'Alta', 'Baja')
print('Primeras 10 etiquetas de ventas:')
print(etiquetas[:10])
print(f'\nVentas altas: {np.sum(etiquetas == "Alta")}')
print(f'Ventas bajas: {np.sum(etiquetas == "Baja")}')
```

```python
# np.clip() — Limitar valores dentro de un rango (clamping)
# Útil para normalización o limitar valores extremos
señal_ruidosa = np.array([-3.2, 0.5, 1.8, -0.9, 4.5, 0.2, -1.1, 2.7, 5.0, -4.0])
señal_limitada = np.clip(señal_ruidosa, -2.0, 2.0)

print('Señal original: ', señal_ruidosa)
print('Señal limitada: ', señal_limitada)
```

```python
# np.cumsum() / np.cumprod() — Suma y producto acumulativo
# Ejemplo: ingresos acumulados durante una semana
ingresos_semana = np.array([1200, 980, 1500, 870, 1350, 2100, 760])
ingresos_acum = np.cumsum(ingresos_semana)

print('Ingresos diarios   :', ingresos_semana)
print('Ingresos acumulados:', ingresos_acum)
```

```python
# np.dot() / @ — Producto matricial
# Ejemplo: cálculo de predicción lineal (regresión)
# y = X @ w  donde X son features y w son los pesos del modelo
X_features = np.array([[1, 2.5, 0.8],
                        [1, 3.1, 1.2],
                        [1, 1.8, 0.5]])

w_pesos = np.array([0.5, 2.0, -1.5])  # bias, peso1, peso2

predicciones = np.dot(X_features, w_pesos)
print('Predicciones del modelo lineal:', np.round(predicciones, 3))
```

```python
# Broadcasting — Operaciones entre arreglos de distinta forma
# Ejemplo: normalización Min-Max de una matriz de datos
datos_brutos = np.array([[100, 200, 300],
                          [50, 400, 150],
                          [80, 120, 500]])

min_col = datos_brutos.min(axis=0)  # mínimo por columna
max_col = datos_brutos.max(axis=0)  # máximo por columna

# Broadcasting aplica min y max automáticamente por columna
datos_normalizados = (datos_brutos - min_col) / (max_col - min_col)

print('Datos originales:')
print(datos_brutos)
print('\nDatos normalizados (Min-Max):')
print(np.round(datos_normalizados, 3))
```

```python
# np.linalg — Álgebra lineal
# np.linalg.inv() — Inversa de matriz
# np.linalg.det() — Determinante
# np.linalg.norm() — Norma

M = np.array([[4.0, 2.0],
              [1.0, 3.0]])

print('Matriz M:')
print(M)
print(f'\nDeterminante        : {np.linalg.det(M):.2f}')
print(f'Norma Frobenius     : {np.linalg.norm(M):.4f}')
print('\nMatriz inversa M⁻¹  :')
print(np.round(np.linalg.inv(M), 4))

# Verificación: M @ M⁻¹ ≈ Identidad
print('\nVerificación M @ M⁻¹ (debe ser identidad):')
print(np.round(M @ np.linalg.inv(M), 1))
```

# 🐼 PARTE 2: Pandas — DataFrames y Métodos

Pandas es la librería fundamental para manipulación y análisis de datos tabulares en Python. Ofrece estructuras de datos flexibles y poderosas herramientas de análisis.

### 3.1 Creación de DataFrames y Series

```python
# pd.Series() — Estructura unidimensional etiquetada
# Ejemplo: temperatura por ciudad colombiana
temp_ciudades = pd.Series(
    data=[29.5, 18.2, 24.0, 17.8, 31.2],
    index=['Bucaramanga', 'Bogotá', 'Medellín', 'Manizales', 'Barranquilla'],
    name='Temperatura_Promedio_C'
)
print('Serie de temperaturas por ciudad:')
print(temp_ciudades)
print(f'\ndtype: {temp_ciudades.dtype}')
```

```python
# pd.DataFrame() — Estructura bidimensional (tabla)
# Ejemplo: dataset de empleados de una empresa
np.random.seed(10)

df_empleados = pd.DataFrame({
    'nombre'      : ['Laura Ríos', 'Jorge Pérez', 'Camila Torres', 'Andrés Ruiz',
                     'Valentina Cruz', 'Miguel Herrera', 'Isabella Gómez', 'Sebastián Mora'],
    'departamento': ['TI', 'Ventas', 'TI', 'RRHH', 'Ventas', 'TI', 'Marketing', 'RRHH'],
    'salario_cop' : [4500000, 3200000, 5100000, 2800000, 3750000, 4800000, 3600000, 2950000],
    'años_empresa': [3, 7, 1, 5, 2, 4, 6, 3],
    'evaluacion'  : [4.2, 3.8, 4.7, 3.5, 4.1, 4.5, 3.9, 3.6],
    'activo'      : [True, True, True, False, True, True, True, False]
})

print('DataFrame de empleados:')
df_empleados
```

### 3.2 Inspección del DataFrame

```python
# .info() — Resumen del DataFrame: tipos, nulos y memoria
print('=== df.info() ===')
df_empleados.info()
```

```python
# .describe() — Estadísticas descriptivas de columnas numéricas
print('=== df.describe() ===')
df_empleados.describe().round(2)
```

```python
# .head(n) / .tail(n) — Primeras/últimas n filas
print('=== df.head(3) — Primeras 3 filas ===')
print(df_empleados.head(3))

print('\n=== df.tail(3) — Últimas 3 filas ===')
print(df_empleados.tail(3))
```

```python
# .shape, .columns, .dtypes, .index
print(f'.shape   → {df_empleados.shape}  ({df_empleados.shape[0]} filas, {df_empleados.shape[1]} columnas)')
print(f'.columns → {list(df_empleados.columns)}')
print(f'.index   → {df_empleados.index.tolist()}')
print('\n.dtypes →')
print(df_empleados.dtypes)
```

### 3.3 Selección y Filtrado de Datos

```python
# .loc[] — Selección por etiquetas (label-based)
print('=== .loc[0:2, ["nombre", "salario_cop"]] ===')
print(df_empleados.loc[0:2, ['nombre', 'salario_cop']])

# .iloc[] — Selección por posición numérica (position-based)
print('\n=== .iloc[::2, :3] — Filas pares, primeras 3 columnas ===')
print(df_empleados.iloc[::2, :3])
```

```python
# Filtrado booleano
# Ejemplo: empleados activos de TI con evaluación > 4.0
filtro = (df_empleados['activo'] == True) & \
         (df_empleados['departamento'] == 'TI') & \
         (df_empleados['evaluacion'] > 4.0)

print('=== Empleados activos de TI con evaluación > 4.0 ===')
print(df_empleados[filtro][['nombre', 'departamento', 'salario_cop', 'evaluacion']])
```

```python
# .query() — Filtrado con cadena de texto (más legible)
resultado = df_empleados.query('salario_cop > 4000000 and años_empresa >= 3')
print('=== Empleados con salario > 4M y >= 3 años ===')
print(resultado[['nombre', 'salario_cop', 'años_empresa']])
```

```python
# .isin() — Filtrar por lista de valores
depts_objetivo = ['TI', 'Marketing']
df_sub = df_empleados[df_empleados['departamento'].isin(depts_objetivo)]
print('=== Empleados de TI o Marketing ===')
print(df_sub[['nombre', 'departamento', 'evaluacion']])
```

### 3.4 Transformación y Creación de Columnas

```python
# .assign() — Crear nuevas columnas (estilo funcional, no modifica el original)
df_emp2 = df_empleados.assign(
    salario_anual = lambda x: x['salario_cop'] * 12,
    nivel_evaluacion = lambda x: pd.cut(x['evaluacion'],
                                         bins=[0, 3.5, 4.0, 4.5, 5.0],
                                         labels=['Regular', 'Bueno', 'Muy Bueno', 'Excelente'])
)

print('Nuevas columnas asignadas:')
print(df_emp2[['nombre', 'evaluacion', 'nivel_evaluacion', 'salario_anual']].to_string())
```

```python
# .apply() — Aplicar función personalizada por fila o columna
def calcular_bono(row):
    """Bono según evaluación y años de empresa"""
    if row['evaluacion'] >= 4.5:
        return row['salario_cop'] * 0.20
    elif row['evaluacion'] >= 4.0:
        return row['salario_cop'] * 0.10
    else:
        return row['salario_cop'] * 0.05

df_empleados['bono'] = df_empleados.apply(calcular_bono, axis=1)
print('Bonos calculados por empleado:')
print(df_empleados[['nombre', 'evaluacion', 'salario_cop', 'bono']].to_string())
```

```python
# .map() — Mapear valores de una Serie usando diccionario o función
mapa_dept = {'TI': 'Technology', 'Ventas': 'Sales',
             'RRHH': 'Human Resources', 'Marketing': 'Marketing'}

df_empleados['dept_en'] = df_empleados['departamento'].map(mapa_dept)
print('Departamentos mapeados al inglés:')
print(df_empleados[['nombre', 'departamento', 'dept_en']].to_string())
```

```python
# Métodos de cadena (.str) — Procesamiento de texto
print('=== Métodos .str ===')
print('Nombres en mayúscula (.str.upper()):')
print(df_empleados['nombre'].str.upper().values)

print('\nExtraer primer nombre (.str.split().str[0]):')
df_empleados['primer_nombre'] = df_empleados['nombre'].str.split().str[0]
print(df_empleados['primer_nombre'].values)

print('\n¿Nombre contiene "a" (.str.contains())?')
print(df_empleados['nombre'].str.contains('a', case=False).values)
```

### 3.5 Agrupación y Agregación

```python
# .groupby() + .agg() — Agrupación con múltiples funciones de agregación
resumen_dept = df_empleados.groupby('departamento').agg(
    num_empleados  = ('nombre', 'count'),
    salario_prom   = ('salario_cop', 'mean'),
    salario_max    = ('salario_cop', 'max'),
    evaluacion_prom= ('evaluacion', 'mean'),
    bono_total     = ('bono', 'sum')
).round(0)

print('=== RESUMEN POR DEPARTAMENTO ===')
resumen_dept
```

```python
# pd.pivot_table() — Tabla dinámica
# Ejemplo: salario promedio y evaluación media por departamento y estado activo
pivot = pd.pivot_table(
    df_empleados,
    values=['salario_cop', 'evaluacion'],
    index='departamento',
    aggfunc={'salario_cop': 'mean', 'evaluacion': 'mean'}
).round(2)

print('=== TABLA DINÁMICA: Salario y Evaluación por Depto ===')
pivot
```

```python
# .value_counts() — Frecuencia de valores únicos
print('=== Distribución por departamento (.value_counts()) ===')
print(df_empleados['departamento'].value_counts())

print('\n=== Porcentaje por departamento (.value_counts(normalize=True)) ===')
print(df_empleados['departamento'].value_counts(normalize=True).round(3) * 100)
```

### 3.6 Manejo de Datos Faltantes

```python
# Crear DataFrame con valores faltantes
import numpy as np

df_ventas = pd.DataFrame({
    'producto'   : ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Audífonos', 'Webcam'],
    'precio'     : [2500000, 45000, np.nan, 890000, np.nan, 120000],
    'stock'      : [10, np.nan, 50, 8, 25, np.nan],
    'categoria'  : ['Computación', 'Periférico', 'Periférico', None, 'Audio', 'Video']
})

print('DataFrame con valores faltantes:')
print(df_ventas)

# .isna() / .isnull() — Detectar nulos
print('\n.isna() — Mapa de valores faltantes:')
print(df_ventas.isna())

print('\n.isna().sum() — Conteo de nulos por columna:')
print(df_ventas.isna().sum())
```

```python
# .fillna() — Rellenar valores nulos
df_ventas_completo = df_ventas.copy()
df_ventas_completo['precio'].fillna(df_ventas['precio'].median(), inplace=True)  # Mediana
df_ventas_completo['stock'].fillna(0, inplace=True)                              # Cero
df_ventas_completo['categoria'].fillna('Sin Categoría', inplace=True)            # Literal

print('DataFrame sin valores faltantes:')
print(df_ventas_completo)
```

```python
# .dropna() — Eliminar filas/columnas con nulos
df_sin_nulos_filas = df_ventas.dropna()          # Eliminar cualquier fila con nulos
df_sin_nulos_thresh = df_ventas.dropna(thresh=3) # Mantener filas con al menos 3 valores válidos

print(f'Original: {df_ventas.shape} | Sin nulos (dropna): {df_sin_nulos_filas.shape} | Umbral 3: {df_sin_nulos_thresh.shape}')
```

### 3.7 Combinar DataFrames

```python
# pd.merge() — Combinar DataFrames como SQL JOIN
df_clientes = pd.DataFrame({
    'cliente_id': [101, 102, 103, 104],
    'nombre': ['Diana Pinto', 'Ricardo Suárez', 'Natalia Vargas', 'Esteban Leal'],
    'ciudad': ['Bogotá', 'Medellín', 'Cali', 'Bucaramanga']
})

df_pedidos = pd.DataFrame({
    'pedido_id' : [1, 2, 3, 4, 5],
    'cliente_id': [101, 102, 101, 103, 105],  # 105 no existe en clientes
    'monto'     : [350000, 120000, 890000, 45000, 670000]
})

# Inner join
df_inner = pd.merge(df_clientes, df_pedidos, on='cliente_id', how='inner')
print('=== INNER JOIN ===')
print(df_inner)

# Left join
df_left = pd.merge(df_clientes, df_pedidos, on='cliente_id', how='left')
print('\n=== LEFT JOIN (todos los clientes, incluso sin pedido) ===')
print(df_left)
```

```python
# pd.concat() — Concatenar DataFrames
df_Q1 = pd.DataFrame({'mes': ['Ene', 'Feb', 'Mar'], 'ventas': [120, 135, 98]})
df_Q2 = pd.DataFrame({'mes': ['Abr', 'May', 'Jun'], 'ventas': [145, 160, 131]})

df_semestre = pd.concat([df_Q1, df_Q2], ignore_index=True)
print('Datos del semestre (concat vertical):')
print(df_semestre)
```

### 3.8 Ordenamiento y Ranking

```python
# .sort_values() — Ordenar por columna(s)
print('=== Top 5 empleados por salario (descendente) ===')
print(df_empleados.sort_values('salario_cop', ascending=False)
      [['nombre', 'departamento', 'salario_cop']].head(5).to_string())

print('\n=== Orden por departamento y evaluación (multicolumna) ===')
print(df_empleados.sort_values(['departamento', 'evaluacion'], ascending=[True, False])
      [['nombre', 'departamento', 'evaluacion']].to_string())
```

# 📚 PARTE 3: NumPy y Pandas en Ciencia de Datos
## Ventajas, Desventajas y Uso en Algoritmos

> *Fuentes consultadas: documentación oficial de NumPy (numpy.org), Pandas (pandas.pydata.org), y recursos académicos de ciencia de datos.*

---

## 🔷 NumPy en Ciencia de Datos

### ✅ Ventajas

| Ventaja | Descripción |
|---------|-------------|
| **Velocidad** | Operaciones vectorizadas escritas en C; hasta 100x más rápido que listas de Python |
| **Eficiencia de memoria** | Almacenamiento contiguo en memoria (arrays homogéneos); menor overhead |
| **Broadcasting** | Operaciones entre arreglos de distintas formas sin loops explícitos |
| **Álgebra lineal** | `np.linalg` ofrece inversas, descomposiciones, autovalores; base para ML |
| **Interoperabilidad** | Base de TensorFlow, PyTorch, Scikit-learn, SciPy; todos consumen ndarray |
| **Operaciones ufunc** | Funciones universales (sin, cos, exp, log) aplicadas elemento a elemento |
| **Reproducibilidad** | `np.random.seed()` garantiza experimentos reproducibles |

### ❌ Desventajas

| Desventaja | Descripción |
|------------|-------------|
| **Solo datos homogéneos** | Un ndarray solo puede contener un tipo de dato; mezclar tipos fuerza dtype=object (lento) |
| **Sin etiquetas** | Los ejes no tienen nombres; dificulta el seguimiento de features en datasets reales |
| **Datos faltantes** | No tiene soporte nativo para NaN en enteros; requiere conversión o enmascaramiento |
| **Curva de aprendizaje** | Indexing avanzado, broadcasting y strides pueden ser confusos al inicio |
| **No escala a Big Data** | Para datasets que no caben en RAM se necesita Dask, CuPy u otras alternativas |

### 🔬 Uso en Algoritmos de ML
- **Regresión lineal**: `X @ w` para predicciones, `np.linalg.lstsq` para mínimos cuadrados
- **Redes neuronales**: Forward/backward pass se implementan como multiplicaciones de matrices
- **Normalización**: `(X - X.mean()) / X.std()` aplicado por broadcasting
- **Distancias**: `np.linalg.norm(a - b)` para KNN y clustering
- **PCA manual**: `np.linalg.eig` sobre la matriz de covarianza

---

## 🔶 Pandas en Ciencia de Datos

### ✅ Ventajas

| Ventaja | Descripción |
|---------|-------------|
| **Datos heterogéneos** | Un DataFrame puede mezclar int, float, str, datetime en distintas columnas |
| **Etiquetas significativas** | Columnas con nombres y índice personalizable; el código se auto-documenta |
| **Manejo de nulos** | `NaN`, `NaT` y `pd.NA` son ciudadanos de primera clase; métodos fillna/dropna |
| **GroupBy potente** | Operaciones split-apply-combine en pocas líneas |
| **I/O versátil** | Lee/escribe CSV, Excel, JSON, SQL, Parquet, HDF5 con una sola línea |
| **Series temporales** | `resample`, `rolling`, `shift` facilitan análisis de tiempo |
| **Integración con visualización** | `.plot()` integrado; compatibilidad directa con Matplotlib, Seaborn, Plotly |

### ❌ Desventajas

| Desventaja | Descripción |
|------------|-------------|
| **Memoria elevada** | DataFrames pueden usar 5-10x más RAM que el archivo CSV original |
| **Velocidad menor que NumPy** | El overhead de etiquetas y tipos mixtos lo hace más lento en cálculos numéricos puros |
| **API inconsistente** | Muchos métodos tienen parámetros redundantes (`inplace`, `axis`); curva de aprendizaje |
| **No para operaciones matriciales** | `.values` o `.to_numpy()` necesarios para pasar a sklearn; rompe el flujo |
| **Problemas con datos grandes** | Sin chunking o Dask, DataFrames > RAM colapsan el proceso |
| **Concatenaciones lentas en loop** | `pd.concat` dentro de un for crea copias repetidamente; antipatrón común |

### 🔬 Uso en el Pipeline de Ciencia de Datos
| Etapa | Uso de Pandas |
|-------|---------------|
| **Carga de datos** | `pd.read_csv()`, `pd.read_excel()`, `pd.read_sql()` |
| **EDA** | `.describe()`, `.value_counts()`, `.corr()`, `.groupby()` |
| **Limpieza** | `.fillna()`, `.dropna()`, `.drop_duplicates()`, `.astype()` |
| **Feature Engineering** | `.assign()`, `.apply()`, `.cut()`, `.get_dummies()` |
| **Entrega a modelos** | `.to_numpy()` → Scikit-learn, `.values` → NumPy |
| **Resultados** | Almacenar predicciones en columnas, exportar con `.to_csv()` |

---

## ⚖️ Comparación Directa

| Criterio | NumPy | Pandas |
|----------|-------|--------|
| Estructura principal | `ndarray` N-dimensional | `DataFrame` (2D) / `Series` (1D) |
| Tipos de datos | Homogéneos | Heterogéneos |
| Etiquetas | No (solo índices numéricos) | Sí (nombres de columna e índice) |
| Datos faltantes | Soporte limitado | Soporte nativo completo |
| Velocidad numérica | ⭐⭐⭐⭐⭐ | ⭐⭐⭐ |
| Facilidad de análisis | ⭐⭐⭐ | ⭐⭐⭐⭐⭐ |
| I/O de archivos | ⭐⭐ | ⭐⭐⭐⭐⭐ |
| Uso típico | Cálculo matricial, ML core | EDA, preprocesamiento, reportes |

---

## 🏁 Conclusión

> **NumPy y Pandas no son competidores, son complementarios.** En un pipeline típico de ciencia de datos:
> - Se usa **Pandas** para cargar, limpiar, explorar y transformar datos tabulares con nombres.
> - Se convierte a **NumPy** (`.to_numpy()`) cuando se necesita eficiencia máxima en cálculos numéricos.
> - Los modelos de **Scikit-learn**, **TensorFlow** y **PyTorch** consumen arrays NumPy internamente.
>
> Dominar ambas librerías es una habilidad fundamental en ciencia de datos e inteligencia artificial.

```python
# Demostración final: Pipeline Pandas → NumPy → Modelo
print('=== PIPELINE CIENCIA DE DATOS ===')
print()

# 1. Datos en Pandas (limpieza y transformación)
df_modelo = df_empleados[df_empleados['activo'] == True][['años_empresa', 'evaluacion', 'salario_cop']].copy()
print('1. DataFrame de entrada (Pandas):')
print(df_modelo.to_string())

# 2. Convertir a NumPy para el modelo
X = df_modelo[['años_empresa', 'evaluacion']].to_numpy()
y = df_modelo['salario_cop'].to_numpy()

print(f'\n2. Features X (NumPy shape {X.shape}):')
print(X)

# 3. Normalización con NumPy
X_norm = (X - X.mean(axis=0)) / X.std(axis=0)
print(f'\n3. Features normalizados (Z-score):')
print(X_norm.round(3))

# 4. Regresión lineal con álgebra lineal NumPy
X_b = np.column_stack([np.ones(len(X_norm)), X_norm])  # Agregar columna de 1s
w = np.linalg.lstsq(X_b, y, rcond=None)[0]  # Solución mínimos cuadrados

print(f'\n4. Pesos del modelo: bias={w[0]:.0f}, w1={w[1]:.0f}, w2={w[2]:.0f}')

# 5. Predicciones
y_pred = X_b @ w
residuals = y - y_pred
rmse = np.sqrt(np.mean(residuals**2))
print(f'\n5. RMSE del modelo: ${rmse:,.0f} COP')

# 6. Guardar resultados en Pandas
df_modelo['salario_predicho'] = y_pred.astype(int)
df_modelo['error'] = residuals.astype(int)
print('\n6. Resultados en DataFrame (Pandas):')
print(df_modelo.to_string())
```
